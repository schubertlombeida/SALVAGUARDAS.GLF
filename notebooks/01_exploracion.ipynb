```python
from pathlib import Path
import os, sys, json
root = Path.cwd() if (Path.cwd() / 'src').exists() else Path.cwd().parent
sys.path.insert(0, str(root))
from src.eda import summarize, figures
archive = Path(os.environ.get('GLF_DATASET_ZIP', str(root / 'data/raw/GLF_Galapagos_Datasets_Iniciales.zip')))
if not archive.is_file():
    raise FileNotFoundError('Configure GLF_DATASET_ZIP con el ZIP autorizado.')
```

```python
metrics = summarize(archive)
print(json.dumps(metrics, ensure_ascii=False, indent=2))
```

```python
figures(metrics, root / 'results/figures')
print('Se generaron seis figuras SVG con agregados del corpus.')
```
