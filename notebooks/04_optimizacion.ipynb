# Workshop S5 — Optimización de hiperparámetros del RAG GLF
Análisis reproducible de sensibilidad. La configuración final ya estaba congelada; esta etapa documenta el comportamiento sin volver a seleccionar parámetros.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..')
marginal = pd.read_csv(ROOT/'results/metrics/partial_dependence_marginal.csv')
importance = pd.read_csv(ROOT/'results/metrics/hyperparameter_importance.csv')
stage1 = pd.read_csv(ROOT/'results/metrics/bm25_grid_stage1.csv')
robust = pd.read_csv(ROOT/'results/metrics/configuraciones_robustas.csv')
stage1
```

## Configuración final congelada

```python
final = {'body_k1':0.8,'body_b':0.2,'heading_k1':1.2,'heading_b':0.3,
         'heading_weight':0.30,'glf_authority_boost':0.15,
         'train_recall_at_5':0.8056,'validation_recall_at_5':0.9091,
         'development_recall_at_5':0.8355}
final
```

## Dependencia marginal

```python
for p in ['k1','b','heading_weight','glf_authority_boost']:
    d = marginal[marginal.parameter==p].sort_values('value')
    plt.figure(figsize=(7,4))
    plt.plot(d.value, d.development_recall_at_5*100, marker='o')
    plt.axhline(80, linestyle='--', linewidth=1)
    plt.title(f'Dependencia marginal: {p}')
    plt.xlabel(p)
    plt.ylabel('Recall@5 medio (%)')
    plt.grid(True, alpha=.25)
    plt.show()
```
